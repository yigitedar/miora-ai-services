# Diabetes classifier (Random Forest)

Predicts diabetes from blood glucose and HbA1c. The data comes from a Kaggle dataset that I filtered and reorganized (20,628 records). The labels follow clear glucose/HbA1c thresholds, so the high accuracy mostly shows that the model learned those thresholds; it is not a clinical result.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import joblib
```

```python
# 1. Veriyi yükle
df = pd.read_csv("balanced_diabetes_dataset.csv")

# 2. Giriş/etiket ayır
X = df[["blood_glucose_level", "HbA1c_level"]]
y = df["diabetes"]
```

```python
# 3. Eğitim/test böl
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

# 4. Modeli eğit
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# 5. Kaydet
joblib.dump(model, "model.pkl")
print("✅ Model 'model.pkl' olarak kaydedildi.")
```

```python
from sklearn.metrics import accuracy_score, classification_report

# 6. Evaluate on the held-out 20% test split
y_pred = model.predict(X_test)
print("Test accuracy:", round(accuracy_score(y_test, y_pred), 4))
print(classification_report(y_test, y_pred, digits=3))
```
